### TREATING GENDER 

```python
df["gender"].value_counts()
```

```python
# Standardize gender values
gender_mapping = {
    "female": "Female",
    "Femle": "Female",
    "FEMALE": "Female",
    "F": "Female",
    "male": "Male",
    "Mlae": "Male",
    "M": "Male",
    "MALE": "Male",
}

df["gender"] = (
    df["gender"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map(gender_mapping)
    .fillna("Prefer Not to Say")
)
```
